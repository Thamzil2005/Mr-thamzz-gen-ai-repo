```python
def add_number(a,b):
    return a+b

print(add_number(10,20))
```

```python
def is_even(n):
    if n % 2 == 0:
        return True
    else:
        return False

print(is_even(10))
```

```python
def square(n):
    return n * n

print(square(5))
```

```python
def max_of_two(a, b):
    if a > b:
        return a
    else:
        return b

print(max_of_two(20, 10))
```
